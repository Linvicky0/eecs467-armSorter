```python
import cv2
from cv_bridge import CvBridge

def create_mock_image_msg(png_path):
    # 1. Load the PNG using OpenCV
    cv_image = cv2.imread(png_path)
    if cv_image is None:
        print(f"Error: Could not find image at {png_path}")
        return None

    # 2. Initialize Bridge
    bridge = CvBridge()

    # 3. Convert OpenCV Mat to ROS 2 Image Message
    # For a standard PNG, the encoding is usually 'bgr8'
    img_msg = bridge.cv2_to_imgmsg(cv_image, encoding="bgr8")
    
    # 4. Add header metadata (important for some ROS nodes)
    img_msg.header.frame_id = "camera_link"
    
    return img_msg


   def convertQtVideoFrame(self):
        """!
        @brief      Converts frame to format suitable for Qt

        @return     QImage
        """

        try:
            frame = cv2.resize(self.VideoFrame, (1280, 720))
            img = QImage(frame, frame.shape[1], frame.shape[0],
                         QImage.Format_RGB888)
            return img
        except:
            return None
```
